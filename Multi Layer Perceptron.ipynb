## part 1

```python
import numpy as np
from numpy import random
import pandas as pd
```

Making x with random numbers

```python
# To generate a large number of samples, we use the random function.
# This function generally generates values within the range of 0 to 1.
# Here, by multiplying the random values by a constant, we scaled them to the specified range of 0 to 10,000.


x_train=random.rand(100)
x_train.sort()
print("x_test has number from ",x_train.min()," to ",x_train.max())
```

Calculate y base on functions blow
* we use numpy functions to make y

```python
# Simple y=2x
y_train1=2*x_train

# y= cos(x)
y_train2=np.cos(2*np.pi* x_train)

# y=arctan(x)
y_train3=np.arctan(2*np.pi*x_train)

# y=arcsinhyperbolic(x)
y_train4=np.arcsinh(2*np.pi*x_train)

```

## MLP

### make a model

* The Sequential model allows us to create models layer-by-layer as we need in a multi-layer perceptron and is limited to single-input, single-output stacks of layers.
* Flatten flattens the input provided without affecting the batch size. For example, If inputs are shaped (batch_size,) without a feature axis, then flattening adds an extra channel dimension and output shape is (batch_size, 1)
* Dense layers are used to make a fully connected model and are the hidden layers.
* The last Dense layer is the output layer which contains 1 neurons that decide function answer.


```python
model1 = tf.keras.Sequential([
    tf.keras.layers.Dense(10, input_shape=(1,), activation='relu'),
    tf.keras.layers.Dense(1, activation='linear')
])
model2 = tf.keras.Sequential([
    tf.keras.layers.Dense(10, input_shape=(1,), activation='relu'),
    tf.keras.layers.Dense(1, activation='linear')
])
model3 = tf.keras.Sequential([
    tf.keras.layers.Dense(10, input_shape=(1,), activation='relu'),
    tf.keras.layers.Dense(1, activation='linear')
])
model4 = tf.keras.Sequential([
    tf.keras.layers.Dense(10, input_shape=(1,), activation='relu'),
    tf.keras.layers.Dense(1, activation='linear')
])
```

### compile a model

* Compile function is used here that involves the use of loss, optimizers, and metrics. Here loss function used is sparse_categorical_crossentropy, optimizer used is adam.

```python
model1.compile(loss='mean_squared_error', optimizer='adam')
model2.compile(loss='mean_squared_error', optimizer='adam')
model3.compile(loss='mean_squared_error', optimizer='adam')
model4.compile(loss='mean_squared_error', optimizer='adam')
```

### fit a model

* Epochs tell us the number of times the model will be trained in forwarding and backward passes.
* Batch Size represents the number of samples, If it’s unspecified, batch_size will default to 32.
* Validation Split is a float value between 0 and 1. The model will set apart this fraction of the training data to evaluate the loss and any model metrics at the end of each epoch. (The model will not be trained on this data)

```python
# y=2*x
model1.fit(x_train, y_train1, epochs=10, batch_size=32)

# y=cos(x)
model2.fit(x_train, y_train2, epochs=10, batch_size=20)

# y=arctag(x)
model3.fit(x_train, y_train3, epochs=10, batch_size=30)

# y=arcsinhyperbolic(x)
model4.fit(x_train, y_train4, epochs=10, batch_size=32)
```

### prediction

```python
x_test =random.rand(100)
x_test.sort()
# y_test

# Simple y=2x
y_test1=2*x_test

# y= cos(x)
y_test2=np.cos(2*np.pi* x_test)

# y=arctan(x)
y_test3=np.arctan(2*np.pi*x_test)

# y=arcsinhyperbolic(x)
y_test4=np.arcsinh(2*np.pi*x_test)

# y predict
y_pred1 = model1.predict(x_test).reshape(100,1)

y_pred2 = model2.predict(x_test).reshape(100,1)

y_pred3 = model3.predict(x_test).reshape(100,1)

y_pred4 = model4.predict(x_test).reshape(100,1)
```

### Errore calculation

* In the context of measuring the accuracy of a model's predictions, a lower value of mean squared error (MSE) is generally better than a higher value.

```python
from sklearn.metrics import mean_squared_error


rmse_score1 = np.sqrt(mean_squared_error(y_test1, y_pred1.reshape(100, )))
rmse_score2 = np.sqrt(mean_squared_error(y_test2, y_pred2.reshape(100, )))
rmse_score3 = np.sqrt(mean_squared_error(y_test3, y_pred3.reshape(100, )))
rmse_score4 = np.sqrt(mean_squared_error(y_test4, y_pred4.reshape(100, )))

print("y1: ",rmse_score1," y2: ",rmse_score1," y3: ",rmse_score1," y4: ",rmse_score1)
```

```python
# y 2
plt.plot(x_test, y_test2,label='Test Data')
plt.plot(x_test, y_pred2,label='Prediction')
plt.title(" cos(x)")
plt.legend()
plt.show()
```

```python
# y 3
plt.plot(x_test, y_test3,label='Training Data')
plt.plot(x_test, y_pred3,label='Prediction')
plt.title(" arctag(x)")
plt.legend()
plt.show()
```

```python
import numpy as np
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Flatten
from keras.callbacks import EarlyStopping
from keras.wrappers.scikit_learn import KerasClassifier
from sklearn.model_selection import cross_val_score

# Load the MNIST dataset
(X_train_full, y_train_full), (X_test, y_test) = mnist.load_data()

# Use only 1000 examples for training
num_train_examples = 1000
X_train_full = X_train_full[:num_train_examples]
y_train_full = y_train_full[:num_train_examples]

# Preprocess the data
X_train_full = X_train_full.astype('float32') / 255.
X_test = X_test.astype('float32') / 255.

# Add noise to the input data
noise_factor = 0.2
X_train_noisy_full = X_train_full + noise_factor * np.random.normal(loc=0.0, scale=1.0, size=X_train_full.shape)
X_train_noisy_full = np.clip(X_train_noisy_full, 0., 1.)

# Define the model
def create_model():
    model = Sequential([
        Flatten(input_shape=(28, 28)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

# Wrap the Keras model with KerasClassifier
model = KerasClassifier(build_fn=create_model, epochs=10, batch_size=32, verbose=0)

# Train and evaluate the model with 5-fold cross-validation
scores = cross_val_score(model, X_train_noisy_full, y_train_full, cv=5, scoring='accuracy', verbose=1)
print(f'Average test accuracy: {np.mean(scores)}')
```

This code defines a create_model function that creates the same neural network model as before, and compiles it with the adam optimizer and sparse_categorical_crossentropy loss. We wrap the Keras model with KerasClassifier, which allows us to use it with scikit-learn's cross_val_score function. We set the number of epochs to 10 and the batch size to 32 in the KerasClassifier constructor. Finally, we perform 5-fold cross-validation on the noisy input data with cross_val_score, and print the average test accuracy over all folds.

The goal of cross-validation is to estimate how well a model will generalize to new, unseen data. The cross-validation score gives an estimate of the model's performance on unseen data, based on the performance on the data used in the cross-validation procedure.

Ideally, the cross-validation score should be as high as possible, indicating that the model generalizes well to new data. However, it is important to keep in mind that the cross-validation score is an estimate and may have some variance depending on the particular split of the data used in the cross-validation procedure.

Therefore, it is not necessarily better for the cross-validation score to be more or less than a certain value. The important thing is to interpret the cross-validation score in the context of the problem and the data, and use it to compare different models and evaluate their performance. In general, a higher cross-validation score indicates better model performance, but the difference between scores may not always be significant.
