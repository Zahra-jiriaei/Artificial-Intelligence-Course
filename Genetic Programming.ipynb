# Genetic Programming

### Introduction

### Genetic Algorithm

The genetic algorithm is generally used for search and optimization problems. This algorithm follows the principles of natural selection. This means that only entities with a greater ability to adapt to their environment can proceed to the next generation.

#### Population

A set of individuals or solutions that are considered for proceeding to the next generation.

#### Genes

The parameters that are used to describe each individual in the population.

#### Chromosome

A string formed by combining multiple genes.

#### Fitness Function

The fitness function assigns a numerical score to each individual based on how well it can survive or perform in the given environment compared with other individuals.

#### Selection

The process of selecting parents from the population. Individuals with higher fitness scores are more likely to be selected as parents.

#### Crossover

The process of combining selected parents to produce new offspring.

#### Mutation

A process in which offspring undergo genetic changes. Mutation is used to maintain diversity among the genes in the population.

### Overall Algorithm

The overall algorithm can be summarized as follows:

1. **Randomly initialize the population**
2. **Calculate the fitness score**
3. **Repeat the following steps until the termination condition is met:**
   3.1. **Select parents from the population**
   3.2. **Generate new offspring**
   3.3. **Apply mutation**
   3.4. **Calculate the fitness scores for the new population**

### Genetic code

```python
# import library
import random
import operator
import math 

# define the inputs and outputs to guess the formula for
inputs = [1, 2, 3, 4, 5, 6, 7, 8, 9]
outputs = [2, 4, 6, 8, 10, 12, 14, 16, 18]
len_inputs=len(inputs)

# define the genetic algorithm parameters
population_size = 100
num_generations = 50
mutation_rate = 0.1

# define the functions and terminals to use in the formula
functions = {
    '+': operator.add,
    '-': operator.sub,
    '*': operator.mul,
    '/': operator.truediv,
    '^':operator.pow,
    'sin': math.sin,
    'cos':math.cos
}

terminals = {
    'inputs': inputs
}


def fitness(individual):
    try:
        formula = ''.join(individual)
        predicted_outputs = [eval(formula, functions, terminals) for x in inputs]
        diff = [abs(outputs[i] - predicted_outputs[i]) for i in range(len(inputs))]
        return 1 / (1 + sum(diff))
    except:
        return 0

# initialize the population
def generate_random_formula(num_vars):
    operators = list(functions.keys())
    return ' '.join(str(random.choice([random.randint(1,10), random.choice(operators)])) for _ in range(num_vars))

#population = [['inputs', '+', '1'] for i in range(population_size)]
population = [generate_random_formula(len_inputs) for i in range(population_size)]

# evolve the population = new generation
# selection and crossover
for generation in range(num_generations):
    # selection
    population = sorted(population, key=fitness, reverse=True)
    best_fit = population[0]
    
    # if there is no erore
    if fitness(best_fit) == 1:
        break
    
    # make new population
    new_population = [best_fit]
    
    #crossover
    while len(new_population) < population_size:
        parent1 = random.choice(population[:50])
        parent2 = random.choice(population[:50])
        child = []
        for i in range(len(parent1)):

# consider mutation
            #  distribute randomly
            if random.random() < mutation_rate:
                child.append(random.choice(list(functions.keys()) + list(terminals.keys())))
            else:
                # i==0
                if i < len(parent2):
                    child.append(parent2[i])
                else:
                    child.append(parent1[i])
        new_population.append(child)
    population = new_population

# print the best fit formula
print(f"Best fit formula: {''.join(best_fit)}")

# test the formula with a new input
new_input = 12
new_output = eval(''.join(best_fit), functions, {'inputs': [new_input]})
print(f"Output for {new_input} is {new_output}")
```

The genetic algorithm was implemented according to the specified requirements.

### Implementation Approach

First, we define the inputs and outputs of the function to be estimated, along with the terminals and operators. In the next step, we define the fitness function so that the score of each individual in the population can be calculated.

We then randomly generate an initial population. As discussed earlier, the genetic algorithm requires a loop to generate successive generations. Since 50 generations were required in this implementation, the loop was executed 50 times.

Within each iteration, the fitness score of each individual is first evaluated, and the individual with the highest score is carried over to the next generation. To select the remaining individuals, two parents are first selected randomly. The next generation is then generated by selecting from higher-fitness parents while applying the specified mutation rate and random selection.

This process is repeated for 50 generations. After completing the iterations, the estimated function is obtained, and its output can be estimated by providing input values.

### Challenges

One of the main challenges was selecting an appropriate function for calculating the fitness score. There are multiple possible functions that could be used for this purpose. After testing several alternatives, the function used in the final implementation was selected.

Some challenges still remain. For example, when the code is executed, the `eval` operation may or may not be performed depending on the function generated in each run. One possible reason for this behavior is that the generated function may not have a valid or sufficiently structured form. Since these functions are generated randomly, the resulting function may not always be valid or suitable for evaluation.

* Initial population size: 100

### Parent Selection

To select the parents, the fitness score of each individual was first calculated using the defined fitness function. The individuals were then sorted according to their fitness scores. Half of the individuals were selected from the initial population, and the parents were randomly selected from this subset.

### Generation of the Next Generation

Fifty generations were created. To generate these generations, the process of calculating the fitness score of each individual, selecting parents, and generating a new generation according to the mutation rate was repeated 50 times.

Mutation was implemented using a randomization-based approach. Several conditions were introduced to maintain diversity and prevent all offspring from having identical characteristics. Depending on these conditions, a new offspring was selected in some cases, while in others, the parent itself was carried over to the next generation.

### Algorithm Termination Condition

In general, the algorithm terminates after generating 50 generations. However, there are additional termination conditions. For example, if the new generation exceeds the size of the population, the loop is not executed. In addition, if a generated function produces a value that exactly matches the target value, the algorithm terminates, and that function is returned as the output.

### Increasing the Mutation Rate


```python
# import library
import random
import operator
import math 
import numpy as np


# define the inputs and outputs to guess the formula for
inputs = 20*np.random.rand(10)
outputs = 3* inputs

len_inputs=len(inputs)

# define the genetic algorithm parameters
population_size = 100
num_generations = 50
mutation_rate = 0.1

# define the functions and terminals to use in the formula
functions = {
    '+': operator.add,
    '-': operator.sub,
    '*': operator.mul,
    '/': operator.truediv,
    '^':operator.pow,
#    'sin': math.sin,
#    'cos':math.cos
}

terminals = {
    'inputs': inputs
}


def fitness(individual):
    try:
        formula = ''.join(individual)
        predicted_outputs = [eval(formula, functions, terminals) for x in inputs]
        diff = [abs(outputs[i] - predicted_outputs[i]) for i in range(len(inputs))]
        return 1 / (1 + sum(diff))
    except:
        return 0

# initialize the population
def generate_random_formula(num_vars):
    operators = list(functions.keys())
    return ' '.join(str(random.choice([random.randint(1,10), random.choice(operators)])) for _ in range(num_vars))

#population = [['inputs', '+', '1'] for i in range(population_size)]
population = [generate_random_formula(len_inputs) for i in range(population_size)]

# evolve the population = new generation
# selection and crossover
for generation in range(num_generations):
    
    # selection
    population = sorted(population, key=fitness, reverse=True)
    best_fit = population[0]
    
    # if there is no erore
    if fitness(best_fit) == 1:
        break
    
    # make new population
    new_population = [best_fit]
    
    #crossover
    while len(new_population) < population_size:
        parent1 = random.choice(population[:50])
        parent2 = random.choice(population[:50])
        child = []
        for i in range(len(parent1)):

# consider mutation
            #  distribute randomly
            if random.random() < mutation_rate:
                child.append(random.choice(list(functions.keys()) + list(terminals.keys())))
            else:
                # i==0
                if i < len(parent2):
                    child.append(parent2[i])
                else:
                    child.append(parent1[i])
        new_population.append(child)
    population = new_population

# print the best fit formula
print(f"Best fit formula: {''.join(best_fit)}")
```
